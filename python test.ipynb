```python
price = int(input("상품 가격(원): "))   # 입력: 숫자를 받아 price에 저장
total = price * 1.1                   # 처리: 부가세 10%를 더한 금액 계산
print("결제 금액:", int(total), "원")   # 출력: 소수점을 버리고 화면에 표시
```

```python
price = int(input("출력 비용: "))     # 출력 비용 입력
add_price = int(input("추가 금액: ")) # 추가 금액 입력

total = price + add_price           # 두 값을 더함 (500 + 200 = 700)
print("결제 금액:", int(total), "원")
```

```python
price = int(input("상품 가격(원): "))

if price <= 0:
    print("올바른 가격을 입력해주세요.")
else:
    total = int(price * 1.1)
    print("결제 금액:", total, "원")
```

```python
print("안녕하세요 철수내 야채입니다.")
print("-----------------------")

# 1. 단가를 미리 정해두기
양파단가 = 1000
토마토단가 = 1500
감자단가 = 800

print("양파단가 =", 양파단가)
print("토마토단가 =", 토마토단가)
print("감자단가 =", 감자단가)
print("-----------------------")

# 2. 개수 입력받기
양파개수 = int(input("필요한 양파 개수 입력 :"))
토마토개수 = int(input("필요한 토마토 개수 입력 :"))
감자개수 = int(input("필요한 감자 개수 입력 :"))

# 3. 부가세 포함 금액 계산하기 (단가 * 개수 * 1.1)
양파금액 = int((양파단가 * 양파개수) * 1.1)
토마토금액 = int((토마토단가 * 토마토개수) * 1.1)
감자금액 = int((감자단가 * 감자개수) * 1.1)

print("-----------------------")

# 4. 계산된 '금액' 출력하기
print("양파 결제 금액 :", 양파금액, "원")
print("토마토 결제 금액 :", 토마토금액, "원")
print("감자 결제 금액 :", 감자금액, "원")
```
